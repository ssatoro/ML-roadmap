# Data Cleaning

```python
import pandas as pd
import numpy as np

raw_df = pd.read_csv('/Users/ftr/ftrs/Code/ML-roadmap/Titanic_project/data/raw/train.csv')
clean_df = raw_df.copy(deep=True)
print(clean_df["Age"])
```

#### Preparation Strategy

The task is to fill in missing values while distorting the original data as little as possible

**Age** - using *mean()*, because there are no significant outliers

**Cabin** - keeping *NaN* values, because there is no way to calculate the mean of cabins or reliably fill in the missing values

**Embarked** - using *mode()*, because there are only 2 missing values and only 3 possible categories

#### Missing Values Treatment

```python
clean_df["Age"] = clean_df["Age"].fillna(clean_df["Age"].mean())
clean_df["Embarked"] = clean_df["Embarked"].fillna(clean_df["Embarked"].mode()[0])
```

#### Data Integrity

```python

print(f'Columns amount before: {raw_df.columns.size}')
print(f'Columns amount after: {clean_df.columns.size}')

print(f"The columns values weren't changed: {all(raw_df["Survived"] == clean_df["Survived"])}")

print(f'Amount of complete duplicate: {clean_df.duplicated().sum()}')

print(f'Number of passes in "Embarked": {clean_df["Embarked"].isna().sum()}')
print(f'Number of passes in "Age": {clean_df["Age"].isna().sum()}')
```

#### Before/After Comparsion

```python
data_integrity_check_df = pd.DataFrame()
data_integrity_check_df["Filled before"] = raw_df.count()
data_integrity_check_df["Filled after"] = clean_df.count()
data_integrity_check_df["Missing before"] = raw_df.isna().sum()
data_integrity_check_df["Missing after"] = clean_df.isna().sum()
data_integrity_check_df["Change"] = raw_df.isna().sum() - clean_df.isna().sum()

print(data_integrity_check_df)
```

#### Cleaning Report

The following problems were identified - the inability to fill in missing values in Cabin without making up new or non-existent cabin names

Decisions made for the problematic features: 
- **Аge** - to fill in missing values with the mean, because the number of outliers is minimal
- **Embarked** - to fill in missing values with the most frequent value, because this approach will distort the original data in categorical fields as little as possible

It was decided to leave missing values only in Cabin to avoid distorting the original data

**Potential problems:** due to the large number of missing values in Age(20%) filling them in with the mean could inaccurately represent the actual data. However, this is the best approach under these circumstances. Filling in missing values in Embarked using mode() should not significantly distort the original data because of the small proportion of missing values(0.22%)
